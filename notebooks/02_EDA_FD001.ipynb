## 2. Load Training Data

```python
import pandas as pd

file_path = "../data/raw/train_FD001.txt"

df = pd.read_csv(
    file_path,
    sep=r"\s+",
    header=None
)

df.head()
```

## 3. Define Column Names

```python
columns = [
    "engine_id",
    "cycle",
    "setting_1",
    "setting_2",
    "setting_3",
    "sensor_1",
    "sensor_2",
    "sensor_3",
    "sensor_4",
    "sensor_5",
    "sensor_6",
    "sensor_7",
    "sensor_8",
    "sensor_9",
    "sensor_10",
    "sensor_11",
    "sensor_12",
    "sensor_13",
    "sensor_14",
    "sensor_15",
    "sensor_16",
    "sensor_17",
    "sensor_18",
    "sensor_19",
    "sensor_20",
    "sensor_21"
]

df.columns = columns

df.head()
```

## 4. Basic Dataset Information

```python
print("Dataset shape:", df.shape)

print("\nData types:")
print(df.dtypes)

print("\nDataset information:")
df.info()
```

## 5. Check Missing Values

```python
missing_values = df.isnull().sum()

print("Missing values per column:")
print(missing_values)

print("\nTotal missing values:", df.isnull().sum().sum())
```

## 8. Engine Lifetime Analysis

```python
#Here, we want to determine how many cycles each engine completed before failure.
engine_lifetime = df.groupby("engine_id")["cycle"].max()

print(engine_lifetime.head())
```

```python
engine_lifetime.describe()
```

## 9. Engine Lifetime Distribution

```python
plt.figure(figsize=(10, 6))

plt.hist(engine_lifetime, bins=20, color='black')

plt.title("Engine Lifetime Distribution")
plt.xlabel("Number of Cycles")
plt.ylabel("Number of Engines")

plt.show()
```

## 10. Sensor Summary Statistics

```python
sensor_columns = [col for col in df.columns if col.startswith("sensor_")]

sensor_summary = df[sensor_columns].describe().T

sensor_summary
```

## 12. Operating Settings Distribution

```python
settings = ["setting_1", "setting_2", "setting_3"]

df[settings].hist(
    figsize=(12, 4),
    bins=30
)

plt.suptitle("Operating Settings Distribution", fontsize=16)

plt.tight_layout()

plt.show()
```

## 13. Check Constant Sensors

```python
sensor_unique_values = df[sensor_columns].nunique()

print(sensor_unique_values)
```

```python
constant_sensors = sensor_unique_values[
    sensor_unique_values == 1
]

print("Constant sensors:")
print(constant_sensors)
```

## 14. Sensor Trajectory for One Engine

```python
#الآن نريد نشوف كيف تتغير Sensors مع مرور الـcycles لمحرك واحد.
engine_1 = df[df["engine_id"] == 1]

engine_1.head()
```

## 15. Plot Selected Sensor Trajectories

```python
selected_sensors = [
    "sensor_2",
    "sensor_4",
    "sensor_7",
    "sensor_11",
    "sensor_15"
]

plt.figure(figsize=(12, 7))

for sensor in selected_sensors:
    plt.plot(
        engine_1["cycle"],
        engine_1[sensor],
        label=sensor
    )

plt.title("Sensor Trajectories - Engine 1")
plt.xlabel("Cycle")
plt.ylabel("Sensor Value")
plt.legend()

plt.show()
```

## 16. Mean Sensor Behavior Over Lifecycle

```python
df["lifecycle_progress"] = (
    df["cycle"] /
    df.groupby("engine_id")["cycle"].transform("max")
)
```

## 17. Plot Mean Sensor Behavior

```python
plt.figure(figsize=(12, 7))

for sensor in selected_sensors:
    mean_sensor = (
        df.groupby("lifecycle_progress")[sensor]
        .mean()
    )

    plt.plot(
        mean_sensor.index,
        mean_sensor.values,
        label=sensor
    )

plt.title("Mean Sensor Behavior Over Lifecycle")
plt.xlabel("Lifecycle Progress")
plt.ylabel("Mean Sensor Value")
plt.legend()

plt.show()
```

## 18. Correlation Analysis

```python
correlation_matrix = df[
    ["cycle"] + sensor_columns
].corr()

correlation_matrix.head()
```

## 19. Correlation Heatmap

```python
plt.figure(figsize=(16, 12))

sns.heatmap(
    correlation_matrix,
    cmap="coolwarm",
    center=0
)

plt.title("Sensor Correlation Heatmap")

plt.show()
```

## 20. Create Training RUL

```python
df["RUL"] = (
    df.groupby("engine_id")["cycle"].transform("max")
    - df["cycle"]
)

df[["engine_id", "cycle", "RUL"]].head(10)
```

## 21. RUL Distribution

```python
plt.figure(figsize=(10, 6))

plt.hist(
    df["RUL"],
    bins=30
)

plt.title("Training RUL Distribution")
plt.xlabel("Remaining Useful Life (Cycles)")
plt.ylabel("Number of Records")

plt.show()
```

## 22. Sensor-RUL Correlation

```python
rul_correlation = (
    df[sensor_columns + ["RUL"]]
    .corr()["RUL"]
    .drop("RUL")
    .sort_values()
)

rul_correlation
```

## 23. Visualize Sensor-RUL Correlation

```python
plt.figure(figsize=(10, 7))

rul_correlation.plot(kind="barh")

plt.title("Sensor Correlation with RUL")
plt.xlabel("Correlation with RUL")
plt.ylabel("Sensor")

plt.show()
```

## 24. Remove Constant Sensors

Constant sensors do not provide useful variation for predicting Remaining Useful Life (RUL).
Therefore, we remove them before building the machine learning models.–

```python
# Identify constant sensors
constant_sensors = sensor_unique_values[
    sensor_unique_values == 1
].index.tolist()

print("Constant sensors:")
print(constant_sensors)

# Remove constant sensors
df_clean = df.drop(columns=constant_sensors)

print("\nOriginal shape:", df.shape)
print("Shape after removing constant sensors:", df_clean.shape)

df_clean.head()
```
